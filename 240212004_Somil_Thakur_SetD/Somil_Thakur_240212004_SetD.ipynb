# Question 1: Word2Vec on Health & Fitness Sentences

In this notebook we train a Word2Vec model on 20 health and fitness sentences using **Gensim**,
and then look at the words the model thinks are similar.

## Step 0: Install and import libraries

```python
!pip install gensim
```

```python
import string
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from gensim.models import Word2Vec

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

## Step 1: Load the sentences into a Python list

```python
sentences = [
    "A balanced diet helps maintain a healthy body weight",
    "Regular workout sessions improve overall muscle strength",
    "Drinking enough water daily supports good digestion",
    "A high protein diet is common among strength athletes",
    "Cardio workouts help improve heart and lung health",
    "Skipping meals can slow down your metabolism over time",
    "Stretching before a workout reduces the risk of injury",
    "A balanced diet should include fruits, vegetables and whole grains",
    "Strength training workouts build muscle and improve bone density",
    "Getting enough sleep is important for muscle recovery",
    "A sudden change in diet can cause digestive discomfort",
    "Yoga is a low-impact workout that improves flexibility",
    "Doctors recommend a balanced diet to prevent lifestyle diseases",
    "Interval training workouts burn calories quickly",
    "A high-sugar diet is linked to several health problems",
    "Consistent workout routines improve long-term fitness levels",
    "Nutritionists often design a diet plan based on individual goals",
    "Warm-up exercises prepare the body for an intense workout",
    "A vegetarian diet can still provide sufficient protein",
    "Rest days are an important part of any workout schedule"
]

print("Total number of sentences:", len(sentences))
```

Save the corpus as a CSV file (needed for the submission).

```python
df = pd.DataFrame({"text": sentences})
df.to_csv("word2vec_corpus.csv", index=False)
df.head()
```

## Step 2: Convert to lowercase and remove punctuation

```python
def clean_text(text):
    text = text.lower()
    for p in string.punctuation:
        text = text.replace(p, " ")   # replace punctuation with a space
    return text

cleaned_sentences = []
for s in sentences:
    cleaned_sentences.append(clean_text(s))

for s in cleaned_sentences[:5]:
    print(s)
```

## Step 3: Tokenize each sentence into words

```python
tokenized_sentences = []
for s in cleaned_sentences:
    tokens = word_tokenize(s)
    tokenized_sentences.append(tokens)

for t in tokenized_sentences[:5]:
    print(t)
```

## Step 4 (Optional): Remove stopwords

```python
stop_words = set(stopwords.words('english'))

final_sentences = []
for tokens in tokenized_sentences:
    new_tokens = []
    for word in tokens:
        if word not in stop_words:
            new_tokens.append(word)
    final_sentences.append(new_tokens)

for t in final_sentences[:5]:
    print(t)
```

```python
# compare the number of unique words before and after stopword removal
words_before = set()
for tokens in tokenized_sentences:
    for w in tokens:
        words_before.add(w)

words_after = set()
for tokens in final_sentences:
    for w in tokens:
        words_after.add(w)

print("Unique words with stopwords   :", len(words_before))
print("Unique words without stopwords:", len(words_after))
```

## Step 5: Train the Word2Vec model

The dataset is very small, so we use small values:
- `vector_size = 50` : each word becomes a list of 50 numbers
- `window = 3` : look at 3 words on the left and right
- `min_count = 1` : keep every word (even if it appears only once)
- `sg = 1` : use the skip-gram method (works better on small data)
- `epochs = 100` : go over the data many times because it is small
- `seed = 42` and `workers = 1` : so that we get the same result every time

```python
model = Word2Vec(
    sentences=final_sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,
    epochs=100,
    seed=42,
    workers=1
)

print("Model trained!")
```

## Step 6: Print the complete vocabulary

```python
vocab = list(model.wv.index_to_key)

print("Total number of sentences:", len(sentences))
print("Vocabulary size:", len(vocab))
print()
print(vocab)
```

## Step 7: Word vector for "diet"

```python
diet_vector = model.wv['diet']

print("Shape of the vector:", diet_vector.shape)
print()
print("First 10 values:")
print(diet_vector[:10])
```

```python
print("Full vector for 'diet':")
print(diet_vector)
```

## Step 8: Top 5 words most similar to "diet"

```python
similar_diet = model.wv.most_similar('diet', topn=5)

print("Top 5 words similar to 'diet':")
for word, score in similar_diet:
    print(word, ":", round(score, 4))
```

## Step 9: Top 5 words most similar to "workout"

```python
similar_workout = model.wv.most_similar('workout', topn=5)

print("Top 5 words similar to 'workout':")
for word, score in similar_workout:
    print(word, ":", round(score, 4))
```

Show both results in a small table using pandas.

```python
table = pd.DataFrame({
    "diet - similar word": [w for w, s in similar_diet],
    "diet - score": [round(s, 4) for w, s in similar_diet],
    "workout - similar word": [w for w, s in similar_workout],
    "workout - score": [round(s, 4) for w, s in similar_workout]
})
table
```

## Step 10: Interpretation

Some of the similar words make sense. For **workout**, the model found *workouts*, *muscle* and *improve*, which really
are related to exercise. For **diet**, words like *workout*, *improve* and *good* come up, which is partly sensible
(diet and workout are both about health), but words like *based* and *enough* are not really related to diet.
This happens because our dataset is very small (only 20 sentences and about 100 unique words), so most words appear
only once and the model does not see enough examples to learn good meanings. Also *diet* and *workout* appear in many
sentences, so they end up close to many common words. With a much bigger dataset the results would be cleaner.

## Questions for Interpretation

**1. Why is the `min_count` parameter needed, and what would happen if you increased it on this dataset?**

`min_count` tells the model to ignore words that appear fewer times than this number. In a big dataset this removes
rare words and spelling mistakes that the model cannot learn well anyway. In our small dataset most words appear
only once, so if we increase `min_count` the vocabulary becomes very small. As the check below shows, the vocabulary
drops from 102 words (min_count=1) to only 14 words (min_count=2) and 7 words (min_count=3), so we lose most of the words.

```python
# small check: vocabulary size for different min_count values
for m in [1, 2, 3]:
    test_model = Word2Vec(sentences=final_sentences, vector_size=50, window=3,
                          min_count=m, sg=1, epochs=100, seed=42, workers=1)
    print("min_count =", m, "-> vocabulary size =", len(test_model.wv.index_to_key))
```

**2. What does increasing the window size change about the relationships the model learns?**

The window is how many words on each side are treated as "context". A small window learns words that appear right
next to each other (more grammar-like / functional similarity). A larger window looks at more of the sentence, so the
model learns broader, topic-level relationships (for example, words from the same kind of sentence about diet or exercise).
In our short sentences a big window covers almost the whole sentence.

**3. Why might your similarity results differ slightly if you re-train the model?**

Word2Vec starts with random vectors and also uses random sampling (negative sampling) and, with several workers,
the order of training can change. So each training run can end up with slightly different vectors and scores.
With such a small dataset this effect is even bigger. Setting `seed` and `workers=1` helps to get the same result again.

# Question 2: Text Preprocessing Pipeline

Cleaning, Tokenization, Stopword Removal, Stemming (Porter & Snowball) and Lemmatization
on noisy health & fitness sentences using **pandas** and **NLTK**.

## Step 0: Import libraries

```python
import re
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize, RegexpTokenizer
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

## Step 1: Load the raw sentences into a DataFrame

```python
raw_sentences = [
    "A balanced DIET helps maintain healthy body weight!!",
    "regular workout sessions improve muscle strength... love it :)",
    "drinking 8 glasses of water daily supports digestion!!",
    "skipping meals can SLOW down your metabolism...",
    "stretching before a workout reduces injury risk!! important",
    "strength training builds muscle & improves bone density",
    "getting 7-8 hrs of sleep is important for recovery :)",
    "yoga is a LOW-impact workout, improves flexibility!!",
    "a high-sugar diet is linked to health problems...",
    "interval training burns calories QUICKLY!! great results",
    "nutritionists design diet plans based on individual goals",
    "rest days are an important part of any workout schedule!!"
]

df = pd.DataFrame({"text": raw_sentences})

# save the dataset as dataset.csv
df.to_csv("dataset.csv", index=False)

df
```

## Step 2: Cleaning function

The function will:
1. make the text lowercase
2. remove punctuation, numbers and special characters (keep only letters and spaces)
3. remove extra spaces

```python
def clean_text(text):
    text = text.lower()                       # 1. lowercase
    text = re.sub(r'[^a-z\s]', ' ', text)     # 2. keep only letters and spaces
    text = re.sub(r'\s+', ' ', text)          # 3. collapse extra spaces
    text = text.strip()
    return text

df["clean_text"] = df["text"].apply(clean_text)
```

Raw and cleaned text side by side:

```python
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 100)
df[["text", "clean_text"]]
```

## Step 3: Tokenization (word_tokenize vs RegexpTokenizer)

```python
# (a) NLTK word_tokenize
df["word_tokens"] = df["clean_text"].apply(word_tokenize)

# (b) RegexpTokenizer that keeps only alphabetic words
regexp_tokenizer = RegexpTokenizer(r'[a-zA-Z]+')
df["regexp_tokens"] = df["clean_text"].apply(regexp_tokenizer.tokenize)

for i in range(3):
    print("Sentence", i + 1, ":", df["clean_text"][i])
    print("word_tokenize   :", df["word_tokens"][i])
    print("RegexpTokenizer :", df["regexp_tokens"][i])
    print()
```

Because we already cleaned the text (no punctuation or numbers left), both tokenizers give the **same** output here.
To see the real difference, let us also try both tokenizers on the **raw** text of the first 3 sentences:

```python
for i in range(3):
    print("Raw sentence", i + 1, ":", df["text"][i])
    print("word_tokenize   :", word_tokenize(df["text"][i]))
    print("RegexpTokenizer :", regexp_tokenizer.tokenize(df["text"][i]))
    print()
```

**Difference:** On raw text, `word_tokenize` keeps punctuation (`!!`, `...`, `:`, `)`) and numbers (`8`) as separate
tokens, while `RegexpTokenizer` with the pattern `[a-zA-Z]+` keeps only the alphabetic words. On cleaned text both are the same.

## Step 4: Stopword removal

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    result = []
    for word in tokens:
        if word not in stop_words:
            result.append(word)
    return result

df["no_stopwords"] = df["word_tokens"].apply(remove_stopwords)

df["count_before"] = df["word_tokens"].apply(len)
df["count_after"] = df["no_stopwords"].apply(len)

df[["clean_text", "count_before", "count_after"]]
```

```python
for i in range(len(df)):
    print(i + 1, df["no_stopwords"][i])
```

## Step 5: Porter Stemmer

```python
porter = PorterStemmer()

def porter_stem(tokens):
    result = []
    for word in tokens:
        result.append(porter.stem(word))
    return result

df["porter"] = df["no_stopwords"].apply(porter_stem)

for i in range(len(df)):
    print(i + 1, df["porter"][i])
```

## Step 6: Snowball Stemmer (English)

```python
snowball = SnowballStemmer('english')

def snowball_stem(tokens):
    result = []
    for word in tokens:
        result.append(snowball.stem(word))
    return result

df["snowball"] = df["no_stopwords"].apply(snowball_stem)

for i in range(len(df)):
    print(i + 1, df["snowball"][i])
```

## Step 7: WordNet Lemmatizer

By default the lemmatizer treats every word as a noun. Verbs like *improves* or *builds* would not change,
so we first try the word as a verb, and if nothing changes we try it as a noun.

```python
lemmatizer = WordNetLemmatizer()

def lemmatize_word(word):
    lemma = lemmatizer.lemmatize(word, pos='v')    # try as a verb
    if lemma == word:
        lemma = lemmatizer.lemmatize(word, pos='n')  # try as a noun
    return lemma

def lemmatize_tokens(tokens):
    result = []
    for word in tokens:
        result.append(lemmatize_word(word))
    return result

df["lemma"] = df["no_stopwords"].apply(lemmatize_tokens)

for i in range(len(df)):
    print(i + 1, df["lemma"][i])
```

## Step 8: Comparison table (Original | Porter | Snowball | Lemma)

We take all distinct words (after stopword removal) and compare the three methods.
The last column shows **DIFFERENT** where Porter and Snowball disagree.

```python
all_words = []
for tokens in df["no_stopwords"]:
    for word in tokens:
        if word not in all_words:
            all_words.append(word)

rows = []
for word in all_words:
    p = porter.stem(word)
    s = snowball.stem(word)
    l = lemmatize_word(word)
    if p == s:
        check = "same"
    else:
        check = "DIFFERENT"
    rows.append([word, p, s, l, check])

compare = pd.DataFrame(rows, columns=["Original Word", "Porter Stem", "Snowball Stem", "Lemma", "Porter vs Snowball"])
compare
```

## Step 9: Vocabulary size at each stage

```python
def vocab_size(column):
    words = set()
    for tokens in df[column]:
        for word in tokens:
            words.add(word)
    return len(words)

print("Raw tokens (after cleaning + tokenizing):", vocab_size("word_tokens"))
print("After stopword removal                  :", vocab_size("no_stopwords"))
print("After Porter stemming                   :", vocab_size("porter"))
print("After Snowball stemming                 :", vocab_size("snowball"))
print("After lemmatization                     :", vocab_size("lemma"))
```

## Step 10: Interpretation

Stemming and lemmatization both reduced the vocabulary a little (63 → 62 words), because our dataset is small and
most words appear in only one form. The big difference is in the quality of the output: the stemmers produced
broken words like *balanc*, *healthi*, *bodi*, *daili*, *quickli* and *schedul*, while the lemmatizer gave real words
like *balance*, *healthy*, *body*, *daily* and *schedule*. Porter and Snowball gave almost the same results, Snowball
was only a little smarter (*quickly → quick*). For a search engine I would prefer **stemming** because it is fast and
it matches many forms of a word together (*improve*, *improves*, *improving* → *improv*). For a classification task or
anything where people read the output, I would prefer **lemmatization** because the words stay meaningful and
easy to understand.

## Questions for Interpretation

**1. Why can stemming produce word forms that are not valid dictionary words, while lemmatization always returns a valid word?**

A stemmer just follows fixed rules that cut off endings like *-ing*, *-es*, *-ly*, *-ity*. It does not check a
dictionary, so the result can be a word that does not exist, for example *daily → daili*, *density → densiti*,
*balanced → balanc*. A lemmatizer looks the word up in the WordNet dictionary and returns its base form (lemma),
so the output is always a real word, for example *improves → improve*, *calories → calorie*.

**2. What is the practical difference between using RegexpTokenizer and NLTK's default word_tokenize, and when would you prefer one over the other?**

`word_tokenize` is a smart tokenizer: it splits punctuation into separate tokens and handles things like
contractions (*don't → do, n't*). `RegexpTokenizer` simply keeps whatever matches the pattern we give it, so with
`[a-zA-Z]+` it returns only words and drops numbers and symbols. I would use `word_tokenize` when punctuation is
useful (for example sentence structure or sentiment like "!!"), and `RegexpTokenizer` when I only want clean words
quickly and want full control over what counts as a token.

**3. Find one word from your dataset where the Porter and Snowball stemmers give different results. Explain why the two algorithms might disagree.**

In our dataset two words give different results:
- **quickly**: Porter → *quickli*, Snowball → *quick*
- **hrs**: Porter → *hr*, Snowball → *hrs*

Snowball (also called "Porter2") is an improved version of the Porter stemmer. It has extra and updated rules,
for example a rule that removes the *-ly* ending (so *quickly* becomes *quick*), while Porter only changes the final
*y* to *i*. Snowball also does not remove the final *s* from very short words like *hrs*, but Porter does.
Because the two algorithms use slightly different rule lists, they sometimes give different stems.
